## Optional local RTG replay

The first run downloads the publicly available template. This starts loopback exchange/trader processes; it does not connect to official Optibook. Keep the flag false if you only want to inspect this notebook.


```python
RUN_REPLAY = False
if RUN_REPLAY:
    subprocess.run([sys.executable, "bot.py", "--engine", "rtg", "--duration", "60", "--speed", "10"], check=True)
else:
    print("Replay skipped. Set RUN_REPLAY = True to start a local match.")
```

## Competition connection

On the competition VM, first use `python3 bot.py --engine optibook --inspect`. Fill and verify `config.json` before using the strategy. `--live` explicitly enables account orders and startup/shutdown cancellation; a dry run does neither. Credentials stay in the provider's local configuration.
